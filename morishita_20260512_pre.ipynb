## 事前課題3

```python
import numpy as np

def check_mean(data, r=0.1, m=10):
    mean_list = []
    for _ in range(m):
        data_sample_mean = data.sample(frac=r).mean()
        mean_list.append(data_sample_mean)

    return np.mean(mean_list), np.std(mean_list)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.1, m=100)
print('----------------- r=0.1 m=100 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.1, m=200)
print('----------------- r=0.1 m=200 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.1, m=300)
print('----------------- r=0.1 m=300 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.2, m=100)
print('----------------- r=0.2 m=100 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.2, m=200)
print('----------------- r=0.2 m=200 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.2, m=300)
print('----------------- r=0.2 m=300 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.3, m=100)
print('----------------- r=0.3 m=100 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.3, m=200)
print('----------------- r=0.3 m=200 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)

mean_area_mean, mean_area_std = check_mean(data=mean_area_col, r=0.3, m=300)
print('----------------- r=0.3 m=300 -----------------')
print('平均:', mean_area_mean)
print('標準偏差:', mean_area_std)
```

試行回数 m を増やすと、標本平均の平均は母平均に近づく。

抽出割合 r を増やすと、標準偏差は小さくなる。

## 事前課題4

**(i)** $X_1, X_2, \ldots, X_n$ が独立に正規分布 $N(\mu, \sigma^2)$ に従うとき、標本平均

$$
\bar{X} = \frac{1}{n}\sum_{i=1}^{n} X_i
$$

は正規分布 $N\left(\mu, \dfrac{\sigma^2}{n}\right)$ に従う。

**(ii)** 確率変数 $X$ が $N(\mu, \sigma^2)$ に従うとき、

$$
Z = \frac{X - \mu}{\sigma}
$$

は標準正規分布 $N(0, 1)$ に従う。

**(iii)** 標準正規分布の上側 $2.5\%$ 点は $1.96$、下側 $2.5\%$ 点は $-1.96$ である。すなわち、

$$
P(-1.96 \leq Z \leq 1.96) = 0.95
$$

---

**導出**

(i) より、標本平均 $\bar{X}$ は平均 $\mu$、分散 $\dfrac{\sigma^2}{n}$（すなわち標準偏差 $\dfrac{\sigma}{\sqrt{n}}$）の正規分布に従う。よって (ii) を $\bar{X}$ に適用すると、

$$
Z = \frac{\bar{X} - \mu}{\sigma / \sqrt{n}}
$$

は標準正規分布 $N(0, 1)$ に従う。

これを (iii) の不等式に代入すると、

$$
-1.96 \leq \frac{\bar{X} - \mu}{\sigma / \sqrt{n}} \leq 1.96
$$

両辺に正の値 $\dfrac{\sigma}{\sqrt{n}}$ を掛けると、

$$
-1.96 \cdot \frac{\sigma}{\sqrt{n}} \leq \bar{X} - \mu \leq 1.96 \cdot \frac{\sigma}{\sqrt{n}}
$$

さらに両辺に $\mu$ を加えると、

$$
\mu - 1.96 \cdot \frac{\sigma}{\sqrt{n}} \leq \bar{X} \leq \mu + 1.96 \cdot \frac{\sigma}{\sqrt{n}}
$$

が得られた。
