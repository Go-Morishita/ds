```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

### ロジスティック回帰

#### 事前課題1の解答

```python
# Iris データセットの読み込み
iris = load_iris()
X = iris.data
y = iris.target

# 全体の30%をテストデータに分割
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
```

#### 事前課題2の解答

**LogisticRegression の概要**

LogisticRegression は、ロジスティック回帰によるクラス分類を行う scikit-learn のクラスである。
- fit()：訓練データから係数を学習する。
- predict()：クラスを予測する。
- predict_proba()：確率値を得る。

単なる対数尤度の最大化ではなく、過学習を防ぐために正則化項を加えた目的関数を最適化しており、その挙動を以下のパラメータで制御できる。

**LogisticRegression のパラメータ**

- l1_ratio：Elastic-Net における L1 正則化の割合を 0 から 1 の実数で指定する。0 なら L2 正則化のみ、1 なら L1 正則化のみと同じになる。
- C：正則化の強さの逆数を指定する。C が小さいほど正則化が強く働き、大きいほど訓練データへの当てはまりが強くなる。
- solver：係数を求める際の最適化アルゴリズムを指定する。


#### 事前課題3の解答

```python
# ロジスティック回帰モデルを学習
lr = LogisticRegression()
lr.fit(X_train, y_train)

# 予測と正解率の計算
y_pred = lr.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"accuracy = {accuracy}")
```

#### 事前課題4の解答

```python
# 予測結果の分割表を出力
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
# 構造の確認
print(iris.target_names)
```

confusion_matrix の出力は、行が実際のクラス、列が予測クラスに対応している。

- setosa と versicolor は、各 15 件すべてが正しく分類された。
- virginica は 15 件中 14 件が正しく分類され、1 件が versicolor と誤分類された。

#### 事前課題5の解答

```python
# 学習したモデルの係数と切片を確認
print(f"features: {iris.feature_names}")
print(f"coefficients:\n{lr.coef_}")
print(f"intercepts: {lr.intercept_}")
```

花びらの大きさに有意性が見られる。
petal length では setosa = -2.29 < versicolor = -0.07817951 < virginica = 2.36967654 となっている。つまり花びらの長さが小さいものは setosa で、大きいものは virginica で、中間が versicolor と解釈できる。
petal width でも同じ傾向が得られるため、花びらの大きさによってモデルが分類していると言える。
