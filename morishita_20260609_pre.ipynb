#### 事前課題3の解答

```python
num_cols = hcv_df.select_dtypes(include='number').columns
imputer = SimpleImputer(strategy='mean')
hcv_imputed = hcv_df.copy()
hcv_imputed[num_cols] = imputer.fit_transform(hcv_df[num_cols])
hcv_imputed.isnull().sum()
```

SimpleImputerはsklearnの欠損値の補完用のクラスである。
パラメータは以下の通りである。
- missing_value：欠損値の指定
- strategy：補完方法の指定。mean（平均）, median（中央値）, most_frequent（最頻値）, constant（定数）がある。
- fill_value：strategy='constant' のとき埋める値を指定する。

#### 事前課題4の解答

```python
iris = load_iris(as_frame=True)
iris_df = iris.data.copy()
iris_df['class'] = iris.target
iris_df.head()
```

```python
iris_df.describe()
```

```python
iris_df['class'].value_counts().sort_index()
```

```python
iris_sample60 = iris_df.sample(n=60, replace=False, random_state=0)
iris_sample60.shape
```

```python
iris_sample60.describe()
```

```python
iris_sample60['class'].value_counts().sort_index()
```

```python
feature_cols = iris.feature_names
mean_compare60 = pd.DataFrame({
    'original_mean': iris_df[feature_cols].mean(),
    'sample60_mean': iris_sample60[feature_cols].mean(),
})
mean_compare60['diff'] = mean_compare60['sample60_mean'] - mean_compare60['original_mean']
mean_compare60
```

```python
class_compare60 = pd.DataFrame({
    'original': iris_df['class'].value_counts(normalize=True).sort_index(),
    'sample60': iris_sample60['class'].value_counts(normalize=True).sort_index(),
})
class_compare60
```

- 60個を非復元抽出した。平均は元データに近いが少しずれる。
- クラスは元が各 1/3（50個ずつ）だが、抽出後は少し偏る。
- 非復元なので同じ個体は重複しない。

#### 事前課題5の解答

```python
iris_sample150 = iris_df.sample(n=150, replace=True, random_state=0)
iris_sample150.shape
```

```python
iris_sample150.index.nunique()
```

```python
iris_sample150.describe()
```

```python
iris_sample150['class'].value_counts().sort_index()
```

```python
mean_compare150 = pd.DataFrame({
    'original_mean': iris_df[feature_cols].mean(),
    'sample150_mean': iris_sample150[feature_cols].mean(),
})
mean_compare150['diff'] = mean_compare150['sample150_mean'] - mean_compare150['original_mean']
mean_compare150
```

```python
class_compare150 = pd.DataFrame({
    'original': iris_df['class'].value_counts(normalize=True).sort_index(),
    'sample150': iris_sample150['class'].value_counts(normalize=True).sort_index(),
})
class_compare150
```

- 150個を復元抽出した。index.nunique() が 97 で、重複ありなのがわかる。
- サンプル数は同じ 150 でも、重複のせいで平均が少しずれる。
- クラスも元の各 1/3 から少し偏る。
