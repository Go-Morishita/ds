```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.preprocessing import StandardScaler
```

## ロジスティック回帰

#### 演習問題1の解答

```python
# 乳がんデータセットの読み込み
data = load_breast_cancer()
X = data.data
y = data.target

# 全体の30%をテストデータに分割
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
```

```python
# ロジスティック回帰モデルを学習
lr = LogisticRegression()
lr.fit(X_train, y_train)
```

デフォルトの100イテレーションで収束しなかったということが警告から確認できる。よって、イテレーションを増加させて収束するかを試す。

```python
lr = LogisticRegression(max_iter=300)
lr.fit(X_train, y_train)
```

```python
lr = LogisticRegression(max_iter=3000)
lr.fit(X_train, y_train)
```

```python
# 予測と評価指標の確認
y_pred = lr.predict(X_test)

print(f"accuracy  = {accuracy_score(y_test, y_pred)}")
print(f"precision = {precision_score(y_test, y_pred, pos_label=0)}")
print(f"recall    = {recall_score(y_test, y_pred, pos_label=0)}")
print(f"f-measure = {f1_score(y_test, y_pred, pos_label=0)}")
```

がんの見逃しが最も重大なので malignant を Positive とした。

```python
# 予測結果の分割表を出力
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

#### 演習問題2の解答

```python
# 説明変数の標準化
scaler = StandardScaler()
X_train_std = scaler.fit_transform(X_train)
X_test_std = scaler.transform(X_test)
```

```python
# ロジスティック回帰モデルを学習
lr_std = LogisticRegression()
lr_std.fit(X_train_std, y_train)
```

```python
# 予測と評価指標の確認
y_pred_std = lr_std.predict(X_test_std)

print(f"accuracy  = {accuracy_score(y_test, y_pred_std)}")
print(f"precision = {precision_score(y_test, y_pred_std, pos_label=0)}")
print(f"recall    = {recall_score(y_test, y_pred_std, pos_label=0)}")
print(f"f-measure = {f1_score(y_test, y_pred_std, pos_label=0)}")
```

```python
# 予測結果の分割表を出力
cm_std = confusion_matrix(y_test, y_pred_std)
print(cm_std)
```

- すべての指標が標準化前と同等以上になり、特に適合率が改善した。
- 誤分類も1件減少している。
- 標準化前はイテレーションを3000まで増やさないと収束しなかったが、標準化後はデフォルト設定のままで収束した。

#### 演習問題4の解答

```python
# 比較用ベースライン：演習問題2のモデル（標準化のみ）の各クラスを基準とした指標
print("--- malignant を Positive とした場合 ---")
print(f"precision = {precision_score(y_test, y_pred_std, pos_label=0)}")
print(f"recall    = {recall_score(y_test, y_pred_std, pos_label=0)}")
print(f"f-measure = {f1_score(y_test, y_pred_std, pos_label=0)}")
print()

print("--- benign を Positive とした場合 ---")
print(f"precision = {precision_score(y_test, y_pred_std, pos_label=1)}")
print(f"recall    = {recall_score(y_test, y_pred_std, pos_label=1)}")
print(f"f-measure = {f1_score(y_test, y_pred_std, pos_label=1)}")
```

```python
# クラス分布の偏りに対応するために class_weight="balanced" を指定して学習
lr_bal = LogisticRegression(class_weight="balanced")
lr_bal.fit(X_train_std, y_train)
```

```python
# テストデータに対する予測と評価
y_pred_bal = lr_bal.predict(X_test_std)

print(f"accuracy = {accuracy_score(y_test, y_pred_bal)}")
print()

print("--- malignant を Positive とした場合 ---")
print(f"precision = {precision_score(y_test, y_pred_bal, pos_label=0)}")
print(f"recall    = {recall_score(y_test, y_pred_bal, pos_label=0)}")
print(f"f-measure = {f1_score(y_test, y_pred_bal, pos_label=0)}")
print()

print("--- benign を Positive とした場合 ---")
print(f"precision = {precision_score(y_test, y_pred_bal, pos_label=1)}")
print(f"recall    = {recall_score(y_test, y_pred_bal, pos_label=1)}")
print(f"f-measure = {f1_score(y_test, y_pred_bal, pos_label=1)}")
```

```python
# 予測結果の分割表を出力
cm_bal = confusion_matrix(y_test, y_pred_bal)
print(cm_bal)
```

クラス分布の偏りに対応するために class_weight="balanced" を指定して学習した。このオプションを指定すると、各クラスの重みが訓練データ中のサンプル数に反比例するように設定される。これにより、少数クラスである malignant の誤分類には大きなペナルティが課されるようになり、学習が多数クラスである benign に偏ることを緩和できると考えた。

分割表から各クラスの再現率を計算すると以下のようになる。
- benign：105 / 107 = 0.981
- malignant：60 / 64 = 0.938

benign の精度を落とさずに、少数クラスである malignant の見逃しは 5 件から 4 件になり、再現率が 0.922 から 0.938 に改善できた。
