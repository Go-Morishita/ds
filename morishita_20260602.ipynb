```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler, MinMaxScaler ,OneHotEncoder
```

#### 演習問題1の回答

```python
breast = load_breast_cancer()
breast_df = pd.DataFrame(breast.data, columns=breast.feature_names)
```

```python
breast_df.head()
```

```python
breast_df.describe()
```

```python
breast_df.boxplot(figsize=(16, 6), rot=90)
plt.title("breast_df")
plt.tight_layout()
plt.show()
```

#### 演習問題2の回答

```python
scaler = StandardScaler()
breast_std = scaler.fit_transform(breast_df)

breast_df_std = pd.DataFrame(breast_std, columns=breast_df.columns)
```

```python
breast_df_std.head()
```

```python
breast_df_std.describe()
```

```python
breast_df_std.boxplot(figsize=(16, 6), rot=90)
plt.title("breast_df_std (StandardScaler)")
plt.tight_layout()
plt.show()
```

#### 演習問題3の回答

```python
scaler_mm = MinMaxScaler()
breast_norm = scaler_mm.fit_transform(breast_df)

breast_df_norm = pd.DataFrame(breast_norm, columns=breast_df.columns)
```

```python
breast_df_norm.head()
```

```python
breast_df_norm.describe()
```

```python
breast_df_norm.boxplot(figsize=(16, 6), rot=90)
plt.title("breast_df_norm (MinMaxScaler)")
plt.tight_layout()
plt.show()
```

#### 演習問題4の回答

**基本統計量に関する考察**
- 標準化後は、各属性の平均がほぼ0、標準偏差がほぼ1になった。
- 正規化後は、各属性の最小値が0、最大値が1になった。

**箱ひげ図に関する考察**
- 標準化後は、標準偏差が1に揃うため、箱の高さのスケール感が属性間で揃う。
- 正規化後は、外枠の範囲が揃うが、外れ値の影響を受けやすいことがわかる。
