#### 事前課題1

#### 事前課題2

```python
scaler = StandardScaler()
iris_std = scaler.fit_transform(iris_df)

iris_df_std = pd.DataFrame(iris_std, columns=iris_df.columns)
```

```python
iris_df_std.head()
```

```python
iris_df_std.describe()
```

```python
iris_df_std.boxplot()
```

#### 事前課題4

**基本統計量に関する考察**
- 標準化後は、各属性の平均がほぼ0、標準偏差がほぼ1になった。
- 正規化後は、各属性の最小値が0、最大値が1になった。

**箱ひげ図に関する考察**
- 標準化後は、標準偏差が1に揃うため、箱の高さのスケール感が属性間で揃う。
- 正規化後は、外枠の範囲が揃うが、外れ値の影響を受けやすいことがわかる。

#### 事前課題5

**Pandas get_dummies()の調査**

曜日などのカテゴリカルデータを、機械学習などで扱える数値データに変換する関数である。ベクトルのようなものが出力で、カテゴリに該当する場所には1を、該当しない場所には0が入る。
これをワンホットエンコーディングという。

引数は以下の通りである。
- columns：対象の列を指定する。
- prefix：生成される列名を指定する。
- drop_first：回帰で発生する多重共線性と呼ばれる現象を抑制する。
- dtype：出力の型を指定する。
