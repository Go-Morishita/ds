```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_diabetes
```

### 単回帰

#### 事前課題1の解答

```python
pearson_df = pd.read_table('data/pearson.txt')
```

```python
pearson_df.describe()
```

```python
pearson_df.plot.box()
```

```python
pearson_df.plot.scatter(x='Father', y='Son')
```

#### 事前課題2の解答

```python
single = pearson_df['Father']
print(type(single))
print(single.shape)
single.head()

double = pearson_df[['Father']]
print(type(double))
print(double.shape)
double.head()
```

pearson_df["Father"] は型が pandas.Series で、pearson_df[["Father"]] は型が pandas.DataFrame である。Series はラベル付きの1次元データ構造で、DataFrame は行と列からなる2次元のデータ構造である。

#### 事前課題3の解答

```python
X = pearson_df[['Father']]
y = pearson_df[['Son']]

reg_lr = LinearRegression()
reg_lr.fit(X, y)

print('Regression coefficient:', reg_lr.coef_)
print('Intercept:', reg_lr.intercept_)
```

#### 事前課題4の解答

```python
print('R^2 score:', reg_lr.score(X, y))
```

#### 事前課題5の解答

```python
plt.scatter(X, y, alpha=0.3, label='data')

plt.plot(X, reg_lr.predict(X), color='red', label='regression line')

plt.xlabel('Father')
plt.ylabel('Son')
plt.legend()
plt.show()
```

### 重回帰

#### 事前課題6の解答

```python
diabetes = load_diabetes()

diabetes_df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
diabetes_df['target'] = diabetes.target

diabetes_df.describe()
```

```python
X = diabetes_df[diabetes.feature_names]
y = diabetes_df['target']

reg_multi = LinearRegression()
reg_multi.fit(X, y)

print('Regression coefficient:', reg_multi.coef_)
print('Intercept:', reg_multi.intercept_)
print('R^2 score:', reg_multi.score(X, y))
```
